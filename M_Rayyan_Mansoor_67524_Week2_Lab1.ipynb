```python
# Count digit in number and their sum
num=int(input("Enter a Number:"))
digit_count=0
digit_sum=0

while num>0:
  digit=num%10
  digit_count=digit_count+1
  digit_sum=digit_sum+digit
  num=num//10
  print("How many digits:",digit_count)
  print("Sum of all digits:",digit_sum)
```
